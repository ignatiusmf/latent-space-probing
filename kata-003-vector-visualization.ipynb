```python
import torch
from matplotlib import pyplot as plt
from rich import print as pprint

# torch.manual_seed(0)

N = 10
n = 2

X = torch.randn(N,n)
Xc = X - X.mean(0, keepdim=True)
pprint(Xc)

for i in range(N):
  plt.quiver(0, 0, Xc[i,0], Xc[i,1], scale=5)

plt.gca() # I HAVE TO SCALE THIS THING SO IT MAKES SENSE VISUALLY
plt.show()


A, B, C = torch.linalg.svd(Xc, full_matrices=False)
pprint('A', A,'B', B,'C',C)
```

```python
k = 1
X_proj = Xc @ C[:, :k]
pprint(X_proj)
X_recon = X_proj @ C[:, :k].T
pprint(X_recon)

for i in range(N):
  plt.quiver(0, 0, X_recon[i,0], X_recon[i,1],scale=5)
  
```

```python
Ck = C[:, :k]
Ckt = Ck.T
pprint(Ck)
pprint(Ckt)

plt.quiver(0, 0, Ck[0], Ck[1], scale=5)
plt.quiver(0,0,Ckt[0,1],Ckt[0,0] , scale=1)
```

```python
import torch
import numpy as np
from matplotlib import pyplot as plt
from rich import print as pprint

np.random.seed(1)

N = 10
n = 3
X = np.random.randn(N,n)
Xc = X - X.mean(0, keepdims=True)

for i in range(N):
  plt.quiver(0, 0, Xc[i, 0], Xc[i, 1], scale=1, angles='xy', scale_units='xy')

pprint(Xc)
plt.gca().set_aspect('equal', adjustable='box')
plt.xlim(-3, 3)
plt.ylim(-3, 3)

plt.show()
```

```python
import torch 
import numpy as np
from matplotlib import pyplot as plt

np.random.seed(0)

N = 10
n = 3
X = np.random.randn(N, n)
Xc = X - X.mean(0, keepdims=True)

for i, v in enumerate(Xc):
  plt.quiver(0, 0, v[0], v[1], scale=1, angles='xy', scale_units='xy') # DAAR MOET NOG GOED DAAR BINNE KOM?
plt.gca().set_aspect('equal', adjustable='box')
plt.xlim(-2.5,2.5)
plt.ylim(-2.5,2.5)

plt.show()

A, B, C = np.linalg.svd(Xc, full_matrices=False)
k = 3
pprint(Xc)
pprint(C)

Xp = Xc @ C
Xr = Xp[:, :k] @ C[:, :k].T

plt.figure(figsize=(5,5))
plt.subplot(1,3,1)
plt.matshow(Xc,fignum=False)

plt.subplot(1,3,2)
plt.matshow(Xp,fignum=False)

plt.subplot(1,3,3)
plt.matshow(Xr,fignum=False)



plt.tight_layout()
plt.show()

```
