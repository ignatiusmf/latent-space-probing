```python
from rich import print
import numpy as np

np.random.seed(0)
N, n = 3, 3
X = np.random.randn(N, n) * 2
Xc = X - X.mean(0, keepdims=True)
A, B, C = np.linalg.svd(Xc, full_matrices=False)

k = 1
Xp = Xc @ C[:, :k]  # (N, n) @ (n, k) = (N, k)
Xr = Xp[:, :k] @ C[:, :k].T  



for v in Xc:
  print(v)

print("Principle components")
for pc in C:
  print(pc)


print("projected vectors")
for pv in Xp:
  print(pv)


print("Reconstructed vectors")
for rv in Xr:
  print(rv)
```
